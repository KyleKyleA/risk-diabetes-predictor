```python
# Author: Kyle Angeles
# File-name: diabetes_data_analysis.py
# Date-Written: 2026/7/8
# Implementing a data analysis and visualization of the diabetes dataset using Python and its libraries.
# through the use of pandas, numpy, and matplotlib, we will explore the dataset and visualize the data to gain insights into the relationships between different features and the target variable (diabetes outcome).
# trying to implement a machine learning model to predict the likelihood of diabetes based on the features in the dataset.
# and coding it for a web application that can be used to predict the likelihood of diabetes based on user input.

# Imported libraries 
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns 
%matplotlib inline 

# Read the diabetes dataset and see what were working with
diabetes_data = pd.read_csv("../data/diabetes_prediction_dataset.csv")
#PRint after sub sample using shape
print('After sub sample: ', diabetes_data.shape)
print('Shape', diabetes_data.shape)
diabetes_data.head()

```

```python
# Data types where working with this particular data set
print('Shape:', diabetes_data.shape)
print('\nDtypes:')
print(diabetes_data.dtypes.value_counts())
diabetes_data.describe(include='all').T.head(20)
```

```python
# Missing values in the dataset
# Note to self everything needs to match and defined correctly
missing = diabetes_data.isnull().sum()
missing_pct = (missing / len(diabetes_data) * 100).round(2)
missing_diabetes_data = pd.DataFrame({'Missing': missing, 'Percent': missing_pct})
missing_diabetes_data = missing_diabetes_data[missing_diabetes_data['Missing'] > 0].sort_values('Missing', ascending=False)
if len(missing_diabetes_data):
    print(missing_diabetes_data)
else:
    print('No missing values')
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
numeric_cols = diabetes_data.select_dtypes(include='number').columns.to_list()
top_numeric = numeric_cols[:15]

corr_matrix = diabetes_data[top_numeric].corr()
plt.figure(figsize=(11,9))
sns.heatmap(corr_matrix, annot=True, fmt='.2f', cmap='coolwarm', center=0, square=True, linewidth=0.5, cbar_kws={'shrink':0.7})
plt.title('Correlation Heatmap for the Early Diabetes Prediction Dataset'); 
plt.tight_layout();
plt.show()
```

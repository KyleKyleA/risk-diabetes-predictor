```python
# Column Description
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
diabetes_data = pd.read_csv("../data/diabetes_prediction_dataset.csv")
print(f"Dataset shape: {diabetes_data.shape[0]} rows, {diabetes_data.shape[1]} columns");
diabetes_data.head()
```

```python
# Visualization Section
import seaborn as sns
import matplotlib.pyplot as plt
numeric_cols = diabetes_data.select_dtypes(include='number').columns.to_list()
top_numeric = numeric_cols[:15]

corr_matrix = diabetes_data[top_numeric].corr()
plt.figure(figsize=(11,9))
sns.heatmap(corr_matrix, annot=True, fmt='.2f', cmap='coolwarm', center=0, square=True, linewidth=0.5, cbar_kws={'shrink':0.7})
plt.title('Correlation Heatmap for the Early Diabetes Prediction Dataset'); 
plt.tight_layout();
plt.show()
```

```python
import seaborn as sns 
import matplotlib.pyplot as plt
sample_data = diabetes_data.sample(min(800, len(diabetes_data)), random_state=42)
g = sns.pairplot(sample_data[['diabetes', 'age', 'bmi']].dropna(), hue='diabetes', diag_kind='kde', height=2.2)
g.figure.suptitle('Age and BMI by Diabetes Outcome', y=1.02);
plt.show()
```

```python
import matplotlib.pyplot as plt 
fig, axes = plt.subplots(1,2, figsize=(13, 4))
if 'blood_glucose_level' in diabetes_data.columns:
    axes[0].scatter(diabetes_data['HbA1c_level'], diabetes_data['blood_glucose_level'], alpha=0.2,s=5, color='steelblue')
    axes[0].set_title('Blood glucose vs HbA1c')
    axes[0].set_xlabel('HbA1c level')
    axes[0].set_ylabel('blood glucose level')

if 'age' in diabetes_data.columns:
    age_group = (diabetes_data['age'] // 5) * 5
    diabetes_data.groupby(age_group)['diabetes'].mean().plot(ax=axes[1], color="tomato")
    axes[1].set_title("Diabetes rate by age")
    axes[1].set_xlabel('age (5-year groups)')
    axes[1].set_ylabel('diabetes rate')
plt.tight_layout();
plt.show()
```

```python
import pandas as pd
missing = pd.DataFrame({
    "Missing Count": diabetes_data.isnull().sum(),
    "Missing %": (diabetes_data.isnull().sum() / len(diabetes_data) * 100).round(2)
})
missing = missing[missing["Missing Count"] > 0].sort_values("Missing %", ascending=False)
print("Columns with missing values: ")
missing 
```

```python
import matplotlib.pyplot as plt 
fix, ax = plt.subplots(figsize=(10, 4))
colors = ["#74c3c" if diabetes_data[col].isnull().any() else "#2ecc71" for col in diabetes_data.columns]
ax.barh(diabetes_data.columns, diabetes_data.isnull().sum(), color=colors)
ax.set_xlabel("Number of missing values")
ax.set_title("Missing values per column")
for i, v in enumerate(diabetes_data.isnull().sum()):
    if v > 0:
        ax.text(v + 5, i, str(v), va="center", fontweight="bold");
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns 
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd 
diabetes_data.head()
numeric_diabetes_data = diabetes_data[["age", "bmi", "HbA1c_level", "blood_glucose_level", "diabetes"]].copy()
numeric_diabetes_data["gender"] = diabetes_data["gender"].map({"male": 0, "female": 1})

corr = numeric_diabetes_data.corr()

fig, ax = plt.subplots(figsize=(10, 8))
mask = np.triu(np.ones_like(corr, dtype=bool))
sns.heatmap(corr, mask=mask, annot=True, fmt=".2f", cmap="RdBu_r", center=0,
            square=True, linewidths=1, ax=ax)
ax.set_title("Correlation Heatmap", fontsize="14", fontweight="bold")
plt.tight_layout()
plt.show()

```
